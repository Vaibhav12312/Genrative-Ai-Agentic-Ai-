````python
 # Top 10 Git Commands to Upload Code on GitHub

''' These are the 10 most important commands you need to upload your local project to GitHub.

## 1. `git init`

Use: Initialize Git in your local project folder.

```
git init
```

Meaning: Start tracking your project using Git.

## 2. `git status`

Use: Check the status of all files.

```
git status
```

Meaning: Shows modified, new and untracked files.

## 3. `git add .`

Use: Add all project files to the staging area.

```
git add .
```

Meaning: Prepare all changes for committing.

## 4. `git commit -m "message"`

Use: Save changes with a message.

```
git commit -m "Initial commit"
```

Meaning: Create a checkpoint of your project.

## 5. `git branch -M main`

Use: Rename your current branch to `main`.

```
git branch -M main
```

Meaning: Set the main branch name.

## 6. `git remote add origin URL`

Use: Connect your local project to your GitHub repository.

```
git remote add origin https://github.com/username/project.git
```

Meaning: Connect local Git with GitHub.

## 7. `git remote -v`

Use: Check the connected GitHub repository.

```
git remote -v
```

Meaning: Verify your remote repository URL.

## 8. `git push -u origin main`

Use: Upload your code to GitHub for the first time.

```
git push -u origin main
```

Meaning: Push your local commits to GitHub and set the upstream branch.

## 9. `git pull`

Use: Download and merge the latest changes from GitHub.

```
git pull
```

Meaning: Update your local project with remote changes.

## 10. `git push`

Use: Upload your latest committed changes to GitHub.

```
git push
```

Meaning: Send new commits to GitHub.

## Complete Practical Workflow

First-time GitHub upload

```
git init
git status
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/project.git
git remote -v
git push -u origin main
```

Every next update

Whenever you make changes to your project:

```
git status
git add .
git commit -m "Updated project"
git push
```

Remember these 3 concepts:

- `git add` → Prepare changes.
- `git commit` → Save changes locally.
- `git push` → Upload saved changes to GitHub. '''
````
